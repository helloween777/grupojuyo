```python
# 1. Importar librerías
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Cargar datos procesados
df = pd.read_csv("../data/features/data6_features.csv")

# 3. Vista general
print(df.shape)
print(df.dtypes)
df.head()

# 4. Estadísticas descriptivas
df.describe()

# 5. Correlaciones
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, cmap="viridis")
plt.title("Matriz de correlación (features)")
plt.show()

# 6. Distribuciones por variable
for col in df.select_dtypes(include="number").columns:
    plt.figure()
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribución de {col}")
    plt.show()

# 7. Observaciones clave
# Aquí puedes anotar manualmente insights sobre variables relevantes, relaciones, etc.
```
