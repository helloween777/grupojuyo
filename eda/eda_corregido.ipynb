```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Configuración visual
sns.set(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)

# Carga de datos
df = pd.read_csv("../data/raw/data6_corregido.csv")  # Ajusta la ruta si es necesario

# Vista general
print("Dimensiones:", df.shape)
display(df.head())
print("\nTipos de datos:")
print(df.dtypes)
print("\nValores nulos:")
print(df.isnull().sum())
print("\nEstadísticas numéricas:")
display(df.describe())

# Histograma de tamaño_pedido
if "tamaño_pedido" in df.columns:
    sns.histplot(df["tamaño_pedido"].dropna(), bins=30, kde=True)
    plt.title("Distribución de tamaño de pedido")
    plt.xlabel("Tamaño de pedido")
    plt.ylabel("Frecuencia")
    plt.show()

# Boxplot de tiempo_entrega por estado_entrega
if "tiempo_entrega" in df.columns and "estado_entrega" in df.columns:
    sns.boxplot(x="estado_entrega", y="tiempo_entrega", data=df)
    plt.title("Tiempo de entrega según estado")
    plt.xlabel("Estado de entrega")
    plt.ylabel("Días")
    plt.xticks(rotation=45)
    plt.show()

# Mapa de calor de correlaciones numéricas
df_num = df.select_dtypes(include="number")
if not df_num.empty:
    sns.heatmap(df_num.corr(), annot=True, cmap="viridis")
    plt.title("Correlación entre variables numéricas")
    plt.show()
```
